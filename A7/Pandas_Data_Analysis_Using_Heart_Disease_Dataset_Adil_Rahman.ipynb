## Question 5: Data Filtering Using Conditions

Apply conditional filtering to:

1.Extract all patients older than 50 years

2.Extract patients who have heart disease and a cholesterol level greater than 240

3.Count how many patients satisfy both conditions

```python
## Apply conditional filtering to: Extract all patients older than 50 years

df[df['age'] > 50]
```

```python
## Apply conditional filtering to: Extract patients who have heart disease and a cholesterol level greater than 240

df[(df['target'] == 1) & (df['chol'] > 240)]
```

```python
## Apply conditional filtering to: Count how many patients satisfy both conditions

# Count of patients who have heart disease and a cholesterol level greater than 240
df[(df['target'] == 1) & (df['chol'] > 240)].shape[0]
```

```python
# Count of patients who are older than 50 years and have heart disease and a cholesterol level greater than 240
df[(df['age'] > 50) & (df['target'] == 1) & (df['chol'] > 240)].shape[0]
```
